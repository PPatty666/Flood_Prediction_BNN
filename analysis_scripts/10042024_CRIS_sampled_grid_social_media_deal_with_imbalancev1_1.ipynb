```python
from tensorflow.keras.layers import Activation, BatchNormalization, Dense, Dropout, InputLayer
from tensorflow.keras.models import Sequential


def make_model_deterministic(n_features):
    model = Sequential()
    model.add(InputLayer(input_shape=(n_features,)))
    model.add(Dropout(0.005))
    model.add(Dense(200, kernel_initializer="glorot_normal"))
    model.add(BatchNormalization())
    model.add(Activation("relu"))
    model.add(Dropout(0.5))
    model.add(Dense(100, kernel_initializer="glorot_normal", use_bias=False))
    model.add(BatchNormalization())
    model.add(Activation("relu"))
    model.add(Dropout(0.25))
    model.add(Dense(50, kernel_initializer="glorot_normal", use_bias=False))
    model.add(BatchNormalization())
    model.add(Activation("relu"))
    model.add(Dropout(0.15))
    model.add(Dense(25, kernel_initializer="glorot_normal", use_bias=False))
    model.add(BatchNormalization())
    model.add(Activation("relu"))
    model.add(Dropout(0.1))
    model.add(Dense(1, activation="sigmoid"))

    model.compile(loss="binary_crossentropy", optimizer="adam", metrics=METRICS)

    return model
```

### Functional API -- Monte Carlo droppout applied

```python
def make_model(n_features):
    inputs = Input(shape=(n_features,))

    x = Dropout(0.05)(inputs, training=True)
    x = Dense(200, kernel_initializer="glorot_normal")(x)
    x = BatchNormalization()(x, training=False)
    x = Activation("relu")(x)
    x = Dropout(0.5)(x, training=False)
    x = Dense(100, kernel_initializer="glorot_normal", use_bias=False)(x)
    x = BatchNormalization()(x, training=False)
    x = Activation("relu")(x)
    x = Dropout(0.25)(x, training=False)
    x = Dense(50, kernel_initializer="glorot_normal", use_bias=False)(x)
    x = BatchNormalization()(x, training=False)
    x = Activation("relu")(x)
    x = Dropout(0.15)(x, training=False)
    x = Dense(25, kernel_initializer="glorot_normal", use_bias=False)(x)
    x = BatchNormalization()(x, training=False)
    x = Activation("relu")(x)
    x = Dropout(0.1)(x, training=False)

    outputs = Dense(1, activation="sigmoid")(x)

    model = Model(inputs=inputs, outputs=outputs)
    model.compile(loss="binary_crossentropy", optimizer="adam", metrics=METRICS)

    return model
```

```python
def plot_metrics(history):
  metrics = ['loss', 'prc', 'precision', 'recall']
  for n, metric in enumerate(metrics):
    name = metric.replace("_"," ").capitalize()
    plt.subplot(2,2,n+1)
    plt.plot(history.epoch, history.history[metric], color=colors[0], label='Train')
    plt.plot(history.epoch, history.history['val_'+metric],
             color=colors[0], linestyle="--", label='Val')
    plt.xlabel('Epoch')
    plt.ylabel(name)
    if metric == 'loss':
      plt.ylim([0, plt.ylim()[1]])
    elif metric == 'auc':
      plt.ylim([0.8,1])
    else:
      plt.ylim([0,1])

    plt.legend()
    plt.show()
```

plot the uncertainty information of the sampled flood risk distribution: standard deviation, the 3rd central moment, the 4th central moment

```python
def plot_uncertainty_moments(
    y_true, y_hat_mc,
    threshold=0.5,
    bins=30,
    share_bins=True,
    is_logits=False,
    show_confusion=True,
    figsize=(10, 8)
):
    """
    Plot MC predictive std, 3rd central moment (mu3), and 4th central moment (mu4)
    split by confusion outcomes (TP/TN/FP/FN).

    Parameters
    ----------
    y_true : array-like, shape (N,)
        Ground truth labels {0,1}.
    y_hat_mc : array-like, shape (S,N,1)
        MC samples of predicted probabilities.
    threshold : float
        Decision threshold applied to MC mean.
    bins : int
        Number of bins (or used to make shared bins).
    share_bins : bool
        If True, use the same bin edges across TP/TN/FP/FN for each metric panel.
    is_logits : bool
        If True, apply sigmoid to y_hat_mc before stats.
    show_confusion : bool
        If True, prints confusion matrix summary.
    figsize : tuple
        Figure size for each 2×2 panel.
    """

    y_true = np.asarray(y_true).astype(int).ravel()
    y_hat_mc = np.asarray(y_hat_mc)

    # --- shape handling: (S,N,1)->(S,N), ensure 2D ---
    if y_hat_mc.ndim == 3 and y_hat_mc.shape[-1] == 1:
        y_hat_mc = y_hat_mc.squeeze(-1)

    if y_hat_mc.ndim != 2:
        raise ValueError(f"Expected 2D MC array after squeezing, got {y_hat_mc.shape}")

    S, M = y_hat_mc.shape
    N = y_true.size

    # Ensure y_hat_mc is (S, N)
    if M == N:
        pass
    elif S == N and M != N:
        y_hat_mc = y_hat_mc.T
        S, M = y_hat_mc.shape
    else:
        raise ValueError(f"Cannot align shapes: y_hat_mc={y_hat_mc.shape}, y_true={y_true.shape}")

    # logits -> probabilities
    if is_logits:
        from scipy.special import expit
        y_hat_mc = expit(y_hat_mc)

    # clamp
    y_hat_mc = np.clip(y_hat_mc, 0.0, 1.0)

    # --- stats per point ---
    p_mean = y_hat_mc.mean(axis=0)          # (N,)
    p_std  = y_hat_mc.std(axis=0)           # (N,)

    diff = y_hat_mc - p_mean                # broadcast (S,N) - (N,)
    mu3  = np.mean(diff**3, axis=0)         # (N,)
    mu4  = np.mean(diff**4, axis=0)         # (N,)

    # --- hard pred from MC mean ---
    y_pred = (p_mean > threshold).astype(int)

    # confusion masks
    TP = (y_true == 1) & (y_pred == 1)
    TN = (y_true == 0) & (y_pred == 0)
    FP = (y_true == 0) & (y_pred == 1)
    FN = (y_true == 1) & (y_pred == 0)

    groups = {
        "True Positives":  TP,
        "True Negatives":  TN,
        "False Positives": FP,
        "False Negatives": FN,
    }

    stds = {k: p_std[m] for k, m in groups.items()}
    m3s  = {k: mu3[m]  for k, m in groups.items()}
    m4s  = {k: mu4[m]  for k, m in groups.items()}

    if show_confusion:
        cm = confusion_matrix(y_true, y_pred)
        print(f"Confusion Matrix @ {threshold:.2f}:\n{cm}")
        print("TP={}, TN={}, FP={}, FN={}".format(
            stds["True Positives"].size,
            stds["True Negatives"].size,
            stds["False Positives"].size,
            stds["False Negatives"].size
        ))

    # --- helpers for shared bins ---
    def make_shared_bin_edges(values_dict):
        finite_vals = np.concatenate([v[np.isfinite(v)] for v in values_dict.values() if v.size])
        if finite_vals.size == 0:
            return np.linspace(0, 1e-6, bins + 1)
        vmin, vmax = float(np.min(finite_vals)), float(np.max(finite_vals))
        if vmin == vmax:
            vmin -= 1e-6
            vmax += 1e-6
        return np.linspace(vmin, vmax, bins + 1)

    def plot_panel(values_dict, xlabel, title):
        bin_edges = make_shared_bin_edges(values_dict) if share_bins else None

        fig, axes = plt.subplots(2, 2, figsize=figsize)
        axes = axes.ravel()

        for ax, (label, arr) in zip(axes, values_dict.items()):
            if arr.size == 0:
                ax.text(0.5, 0.5, "No samples", ha="center", va="center", transform=ax.transAxes)
            else:
                ax.hist(arr, bins=(bin_edges if share_bins else bins), edgecolor="black")
            ax.set_title(f"{label} (n={arr.size})")
            ax.set_xlabel(xlabel)
            ax.set_ylabel("Count")

        fig.suptitle(f"{title} by outcome @ threshold={threshold:.2f}")
        fig.tight_layout()
        plt.show()
        return fig

    # --- plot the three panels ---
    fig_std = plot_panel(stds, r"Predictive $\sigma$ (MC)", r"MC Predictive Uncertainty ($\sigma$)")
    fig_m3  = plot_panel(m3s,  r"3rd central moment ($\mu_3$)", r"MC 3rd Central Moment ($\mu_3$)")
    fig_m4  = plot_panel(m4s,  r"4th central moment ($\mu_4$)", r"MC 4th Central Moment ($\mu_4$))")

    # return arrays + figs for reuse
    out = {
        "p_mean": p_mean,
        "p_std": p_std,
        "mu3": mu3,
        "mu4": mu4,
        "y_pred": y_pred,
        "groups": groups,
        "stds_by_outcome": stds,
        "mu3_by_outcome": m3s,
        "mu4_by_outcome": m4s,
        "figures": (fig_std, fig_m3, fig_m4),
    }
    return out
```

```python
# Read the shapefile: these polygons were digitized from geocoded flooded images and their estimated flood extents.
# The image locations were obtained through post-training community engagement,
# and the polygons are used to validate the model predictions.
flood_img2polygon = gpd.read_file('/content/drive/MyDrive/social_media_based_model/adds_data_run_model070424/flooding_add1.shp')

# Function to extract x and y coordinates from the geometries
def extract_coords(geometry):
    if geometry.geom_type == 'Polygon':
        exterior_coords = geometry.exterior.coords.xy
        return list(exterior_coords[0]), list(exterior_coords[1])
    elif geometry.geom_type == 'MultiPolygon':
        xs, ys = [], []
        for polygon in geometry:
            exterior_coords = polygon.exterior.coords.xy
            xs.append(list(exterior_coords[0]))
            ys.append(list(exterior_coords[1]))
        return xs, ys

# Extract coordinates for all polygons
polygon_xs, polygon_ys = [], []
for geom in flood_img2polygon.geometry:
    xs, ys = extract_coords(geom)
    polygon_xs.append(xs)
    polygon_ys.append(ys)

# Prepare the data for ColumnDataSource
flood_img2polygon['polygon_xs'] = polygon_xs
flood_img2polygon['polygon_ys'] = polygon_ys
```
